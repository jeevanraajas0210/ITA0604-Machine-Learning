```python
import os
os.environ["TF_CPP_MIN_LOG_LEVEL"] = "3"

import numpy as np
import tensorflow as tf
from tensorflow.keras.models import Sequential
from tensorflow.keras.layers import Dense

tf.random.set_seed(42)
np.random.seed(42)

X = np.array([
    [1, 40],
    [2, 50],
    [3, 55],
    [4, 65],
    [5, 70],
    [6, 75],
    [7, 80],
    [8, 90]
], dtype=float)

y = np.array([
    0,
    0,
    0,
    0,
    1,
    1,
    1,
    1
])

X[:,0] = X[:,0] / 8
X[:,1] = X[:,1] / 100

model = Sequential([
    Dense(4, activation="relu", input_shape=(2,)),
    Dense(1, activation="sigmoid")
])

model.compile(
    optimizer="adam",
    loss="binary_crossentropy",
    metrics=["accuracy"]
)

model.fit(X, y, epochs=300, verbose=0)

loss, accuracy = model.evaluate(X, y, verbose=0)

study_hours = 6
attendance = 80

new_student = np.array([[study_hours/8, attendance/100]])

prediction = model.predict(new_student, verbose=0)

print("Accuracy:", round(accuracy*100,2), "%")
print("Prediction Probability:", round(float(prediction[0][0]),4))

if prediction[0][0] >= 0.5:
    print("Result: Student will PASS")
else:
    print("Result: Student will FAIL")
```
