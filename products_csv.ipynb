```python
import numpy as np

prices = np.random.randint(100, 5001, 20)

print("Product Prices:")
print(prices)
```

```python
print("First 5 Prices:")
print(prices[:5])

print("Last 5 Prices:")
print(prices[-5:])
```

```python
gst_prices = prices * 1.18

print("Prices after 18% GST:")
print(np.round(gst_prices, 2))
```

```python
above_2500 = gst_prices[gst_prices > 2500]

print("Products costing more than ₹2500:")
print(np.round(above_2500, 2))
```

```python
between_1000_3000 = gst_prices[
    (gst_prices >= 1000) & (gst_prices <= 3000)
]

print("Products between ₹1000 and ₹3000:")
print(np.round(between_1000_3000, 2))
```

```python
print("Minimum Price:", round(np.min(gst_prices), 2))
print("Maximum Price:", round(np.max(gst_prices), 2))
```

```python
print("Mean:", round(np.mean(gst_prices), 2))
print("Median:", round(np.median(gst_prices), 2))
print("Variance:", round(np.var(gst_prices), 2))
print("Standard Deviation:", round(np.std(gst_prices), 2))
```

```python
total = np.sum(gst_prices)

print("Total Value of All Products:", round(total, 2))
```

```python
average = np.mean(gst_prices)

count = np.sum(gst_prices > average)

print("Average Price:", round(average, 2))
print("Number of Products Above Average:", count)
```

```python
np.savetxt(
    "products.csv",
    gst_prices,
    delimiter=",",
    header="Final_Product_Price",
    comments=""
)

print("File saved successfully as products.csv")
```
