```python
import numpy as np

prices = np.array([100, 200, 300, 400, 500])

print("Mean:", np.mean(prices))
```

```python
import numpy as np

prices = np.array([100, 200, 300, 400, 500])

print("Median:", np.median(prices))
```

```python
from scipy import stats

data = [10, 20, 20, 20, 30, 40]

print("Mode:", stats.mode(data, keepdims=True).mode[0])
```

```python
import numpy as np

prices = np.array([100, 200, 300, 400, 500])

print("Minimum:", np.min(prices))
```

```python
import numpy as np

prices = np.array([100, 200, 300, 400, 500])

print("Maximum:", np.max(prices))
```

```python
import numpy as np

prices = np.array([100, 200, 300, 400, 500])

range_value = np.max(prices) - np.min(prices)

print("Range:", range_value)
```

```python
import numpy as np

data = np.array([10, 30, 50, 70, 90])

print("Variance:", np.var(data))
```

```python
import numpy as np

data = np.array([10, 30, 50, 70, 90])

print("Standard Deviation:", round(np.std(data), 2))
```

```python
import numpy as np

data = np.array([10, 20, 30, 40, 50])

print("25th Percentile:", np.percentile(data, 25))
print("50th Percentile:", np.percentile(data, 50))
print("75th Percentile:", np.percentile(data, 75))
```
