```python
X = [-3, -2, -1, 1, 2, 3]

y = [-1, -1, -1, 1, 1, 1]

weight = 0
mistakes = 0

print("CAT = -1")
print("DOG = +1")
print("--------------------------")

for i in range(len(X)):

    score = weight * X[i]

    if score >= 0:
        prediction = 1
    else:
        prediction = -1

    actual_name = "DOG" if y[i] == 1 else "CAT"
    predicted_name = "DOG" if prediction == 1 else "CAT"

    print("\nAnimal", i + 1)
    print("Actual     :", actual_name)
    print("Prediction :", predicted_name)

    if prediction != y[i]:

        mistakes += 1

        print("MISTAKE")

        weight = weight + y[i] * X[i]

        print("Model learned! New weight =", weight)

    else:
        print("CORRECT")

print("\n========================")
print("TOTAL MISTAKES =", mistakes)
print("========================")
```
