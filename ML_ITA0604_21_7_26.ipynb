```python
import numpy as np
```

```python
x = np.array([-3, -2, -1, 0, 1, 2, 3])
```

```python
output = np.tanh(x)

print("Input Values:")
print(x)

print("Tanh Output:")
print(output)
```

```python
print("\nActivation Function: ReLU")

x = np.array([-5, -2, 0, 3, 7])

relu_output = np.maximum(0, x)

print("Input Values:")
print(x)

print("ReLU Output:")
print(relu_output)
```

```python
print("\nActivation Function: SIGMOID")

x = np.array([-5, 2, 0, 2, 5])

sigmoid_output = 1 / (1 + np.exp(-x))

print("Input Values:")
print(x)

print("Sigmoid Output:")
print(sigmoid_output)
```

```python
print("\nActivation Function: SOFTMAX")

marks = np.array([85, 70, 60])

softmax = np.exp(marks) / np.sum(np.exp(marks))

print("Scores:", marks)
print("Softmax:", softmax)
```

```python

from tensorflow.keras.models import Sequential
from tensorflow.keras.layers import Dense, Dropout


model = Sequential([
    Dense(128, activation='relu', input_shape=(10,)),
    Dropout(0.5),
    Dense(64, activation='relu'),
    Dropout(0.3),
    Dense(1, activation='sigmoid')
])

model.summary()

print("Solution: Dropout reduces overfitting.")
```

```python
from tensorflow.keras.models import Sequential
from tensorflow.keras.layers import Dense

# Larger neural network
model = Sequential([
    Dense(64, activation='relu', input_shape=(10,)),
    Dense(32, activation='relu'),
    Dense(1, activation='sigmoid')
])

model.compile(
    optimizer='adam',
    loss='binary_crossentropy',
    metrics=['accuracy']
)

model.summary()

print("Solution: More neurons, more layers, and more training.")
```

```python
# Single Input Perceptron Example


x = 80


w = 0.5


b = 5
theta = 40


weighted_sum = (x * w) + b

print("Input (x):", x)
print("Weight (w):", w)
print("Bias (b):", b)
print("Threshold (θ):", theta)
print("Weighted Sum (Σ):", weighted_sum)


if weighted_sum >= theta:
    output = 1
else:
    output = 0

print("Output:", output)


if output == 1:
    print("Prediction: PASS")
else:
    print("Prediction: FAIL")
```

```python
from tensorflow.keras.models import Sequential
from tensorflow.keras.layers import Dense
import numpy as np

# Training data
X = np.array([
    [80, 90, 5],
    [60, 70, 3],
    [40, 50, 1],
    [90, 95, 6],
    [30, 40, 1],
    [75, 80, 4]
], dtype=float)

# Normalize input
X = X / 100

# Output (1 = PASS, 0 = FAIL)
y = np.array([1, 1, 0, 1, 0, 1])

# Build Multi-Layer Perceptron
model = Sequential([
    Dense(8, activation='relu', input_shape=(3,)),
    Dense(4, activation='relu'),
    Dense(1, activation='sigmoid')
])

# Compile
model.compile(
    optimizer='adam',
    loss='binary_crossentropy',
    metrics=['accuracy']
)

# Train
model.fit(X, y, epochs=100, verbose=1)

# Test data
test = np.array([[85, 90, 5]]) / 100

prediction = model.predict(test)

print("Prediction Value:", prediction[0][0])

if prediction >= 0.5:
    print("Prediction: PASS")
else:
    print("Prediction: FAIL")
```

```python
from tensorflow.keras.models import Sequential
from tensorflow.keras.layers import Dense
import numpy as np

X = np.array([
    [80, 90, 5],
    [60, 70, 3],
    [40, 50, 1],
    [90, 95, 6],
    [30, 40, 1],
    [75, 80, 4]
], dtype=float)

X = X / 100

y = np.array([1, 1, 0, 1, 0, 1])

model = Sequential([
    Dense(8, activation='relu', input_shape=(3,)),
    Dense(4, activation='relu'),
    Dense(1, activation='sigmoid')
])

model.compile(
    optimizer='adam',
    loss='binary_crossentropy',
    metrics=['accuracy']
)

model.fit(X, y, epochs=100, verbose=0)

loss, accuracy = model.evaluate(X, y, verbose=0)

test = np.array([[85, 90, 5]]) / 100

prediction = model.predict(test, verbose=0)

print("Accuracy:", round(accuracy * 100, 2), "%")
print("Prediction Value:", round(float(prediction[0][0]), 4))

if prediction[0][0] >= 0.5:
    print("Prediction: PASS")
else:
    print("Prediction: FAIL")
```
